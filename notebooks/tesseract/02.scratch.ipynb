```python
from pathlib import Path
from babel import languages
```

```python
_PATH_TESSERACT_MODEL = Path("/opt/homebrew/Cellar/tesseract/5.5.1/share/tessdata/")
```

```python
_TESSARCT_MODEL_FILES = list(_PATH_TESSERACT_MODEL.glob("*.traineddata"))
```

```python
indic_langcodes = ['asm', 'ben', 'brx', 'doi', 'eng', 'gom', 'guj', 'hin', 'kan', 'kas', 'mai', 'mal', 'mar', 'mni', 'nep', 'ori', 'pan', 'san', 'sat', 'snd', 'tam', 'tel', 'urd']
```

```python
for indic_code in indic_langcodes:
    for tess_model_fp in _TESSARCT_MODEL_FILES:
        model_name = tess_model_fp.name.split(".")[0]
        if indic_code == model_name:
            print(f"{indic_code} | {model_name}")
```

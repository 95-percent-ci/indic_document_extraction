# 3. Loading WikiText Provide Data

```python
# path 
# ToDo make it os independent
_PATH_WIKI_SAMPLE = Path("raw_data/sample_wiki/combined_languages_V2.parquet")
df_wiki_sample = pd.read_parquet(_PATH_WIKI_SAMPLE)
```

```python
dogri_ds = pd.read_csv(".temp/dogri_samples.csv").rename(columns=({"SRC":"url",                                                                   "TEXT":"text"}))
bodo_ds = pd.read_csv(".temp/bodo_samples.csv").rename(columns=({"SRC":"url", 
                                                                 "TEXT":"text"}))
bodo_ds['lang_code'] = 'brx'
dogri_ds['lang_code'] = 'doi'
dogri_ds['word_count'] = dogri_ds['text'].apply(lambda x: len(x.split(" ")))
bodo_ds['word_count'] = bodo_ds['text'].apply(lambda x: len(x.split(" ")))
```

```python
COL_REQ_ = ['url', 'text', 'lang_code', 'word_count']
new_data = pd.concat([dogri_ds[COL_REQ_], bodo_ds[COL_REQ_]], axis=0).reset_index(drop=True)
```

```python
data_final = pd.concat([df_wiki_sample[COL_REQ_], new_data], axis=0).reset_index(drop=True)
```

```python
data_final.to_parquet("raw_data/indic_lang_samples/indic_sample.parquet", index=False)
```

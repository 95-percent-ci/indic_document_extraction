```python
import json
```

```python
def read_json(path: str) -> dict:
    """Reads JSON and properly decodes Indic text"""
    with open(path, 'r', encoding='utf-8') as file:
        data = json.load(file)
        if 'header' in data:
            data['header'] = data['header']
        if 'full_text' in data:
            data['full_text'] = data['full_text']
        return data
```
